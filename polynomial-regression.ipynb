```python
df=pd.read_csv('/kaggle/input/polynomial-regression/Ice_cream selling data.csv')
```

```python
X=df.drop('Ice Cream Sales (units)',axis=1)
y=df['Ice Cream Sales (units)']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
from sklearn.preprocessing import PolynomialFeatures
model= PolynomialFeatures(degree=2)
x_tr_poly=model.fit_transform(x_train)
x_ts_poly=model.fit_transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_tr_poly,y_train)
```

```python
y_pred=model.predict(x_ts_poly)
```

```python
model.score(x_ts_poly,y_test)
```

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
mse=mean_squared_error(y_test,y_pred)
mse
```

```python
mae=mean_absolute_error(y_test,y_pred)
mae
```

```python
rmse=np.sqrt(mse)
rmse
```

```python
score=r2_score(y_test,y_pred)
score
```
